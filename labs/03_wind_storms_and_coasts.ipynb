# Lab 3 - Wind, Storms, and Coastal Sea Level

Use wind to push water, create coastal setup, and compare with storm-surge examples.

```python
import numpy as np
import matplotlib.pyplot as plt
from shallowwater import ModelParams, make_grid, compute_dt_cfl, run_model, setup_initial_state, wind_gyre_forcing
```

```python
Nx, Ny = 96, 96
grid = make_grid(Nx, Ny, 1.5e6, 1.5e6)
params = ModelParams(H=200.0, g=9.81, f0=8e-5, beta=1.6e-11, r=1/(5*86400), linear=True)
dt = compute_dt_cfl(grid, params, cfl=0.4)

def forcing(t, g, p):
    return wind_gyre_forcing(t, g, p, tau0=0.08)

def ic(g, p):
    return setup_initial_state(g, p, mode='rest')

out = run_model(tmax=15*86400, dt=dt, grid=grid, params=params, forcing_fn=forcing, ic_fn=ic, save_every=48, out_vars=('eta',))
plt.imshow(out['eta'][-1], origin='lower', extent=[0, grid.Lx/1000, 0, grid.Ly/1000])
plt.colorbar(label='eta [m]')
plt.title('Sea level after wind forcing')
plt.xlabel('x [km]')
plt.ylabel('y [km]')
```
