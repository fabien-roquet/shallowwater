# Lab 5 - Ocean Pathways

Compare several ways to move water: wind, tides, source-sink forcing, and reduced gravity.

```python
import numpy as np
import matplotlib.pyplot as plt
from shallowwater import ModelParams, make_grid, compute_dt_cfl, run_model, setup_initial_state, stommel_arons_forcing
```

```python
Nx, Ny = 96, 96
grid = make_grid(Nx, Ny, 2.0e6, 2.0e6)
params = ModelParams(H=1000.0, g=9.81, f0=8e-5, beta=2e-11, r=1/(10*86400), linear=True)
dt = compute_dt_cfl(grid, params, cfl=0.4)

def forcing(t, g, p):
    return stommel_arons_forcing(t, g, p, Q0=1e-6, R=1.5e5, time_ramp=2*86400)

def ic(g, p):
    return setup_initial_state(g, p, mode='rest')

out = run_model(tmax=20*86400, dt=dt, grid=grid, params=params, forcing_fn=forcing, ic_fn=ic, save_every=60, out_vars=('eta',))
plt.imshow(out['eta'][-1], origin='lower', extent=[0, grid.Lx/1000, 0, grid.Ly/1000])
plt.colorbar(label='eta [m]')
plt.title('Sea-level pattern from source-sink forcing')
plt.xlabel('x [km]')
plt.ylabel('y [km]')
```
