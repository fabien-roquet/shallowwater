# Lab 1 - Waves in a Box

Run a surface bump and a simple seiche. Start with the small experiment below, then compare with the full tutorial notebooks.

```python
import numpy as np
import matplotlib.pyplot as plt
from shallowwater import ModelParams, make_grid, setup_initial_state, compute_dt_cfl, run_model, zero_forcing
```

```python
Nx, Ny = 96, 48
Lx, Ly = 1.2e6, 6.0e5
grid = make_grid(Nx, Ny, Lx, Ly)
params = ModelParams(H=500.0, g=9.81, f0=0.0, beta=0.0, r=0.0, linear=True)
dt = compute_dt_cfl(grid, params, cfl=0.45)
print(f'dt = {dt:.1f} s, wave speed = {np.sqrt(params.g*params.H):.1f} m/s')
```

```python
def ic_bump(g, p):
    return setup_initial_state(g, p, mode='gaussian_bump', amp=0.2, x0=0.35*g.Lx, y0=0.5*g.Ly, R=7.5e4)

out = run_model(tmax=6*3600, dt=dt, grid=grid, params=params, forcing_fn=zero_forcing, ic_fn=ic_bump, save_every=10, out_vars=('eta',))
eta_max = np.max(np.abs(out['eta']), axis=(1,2))
plt.plot(out['t']/3600, eta_max)
plt.xlabel('time [hours]')
plt.ylabel('max |eta| [m]')
plt.title('How large is the surface displacement?')
```
