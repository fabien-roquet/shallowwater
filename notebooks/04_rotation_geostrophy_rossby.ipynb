# Lab 4 - Rotation Changes Everything

Compare a bump on a non-rotating planet with the same bump on a rotating planet.

```python
import numpy as np
import matplotlib.pyplot as plt
from shallowwater import ModelParams, make_grid, setup_initial_state, compute_dt_cfl, run_model, zero_forcing
```

```python
Nx, Ny = 96, 96
grid = make_grid(Nx, Ny, 1.5e6, 1.5e6)

def run_case(f0):
    params = ModelParams(H=500.0, g=9.81, f0=f0, beta=0.0, r=1/(20*86400), linear=True)
    dt = compute_dt_cfl(grid, params, cfl=0.4)
    def ic(g, p):
        return setup_initial_state(g, p, mode='gaussian_bump', amp=0.2, R=1.2e5)
    return run_model(tmax=4*86400, dt=dt, grid=grid, params=params, forcing_fn=zero_forcing, ic_fn=ic, save_every=40, out_vars=('eta',))

out_no_rotation = run_case(0.0)
out_rotation = run_case(1e-4)
print(out_no_rotation['eta'].shape, out_rotation['eta'].shape)
```

```python
plt.imshow(out_rotation['eta'][-1], origin='lower', extent=[0, grid.Lx/1000, 0, grid.Ly/1000])
plt.colorbar(label='eta [m]')
plt.title('Final eta with rotation')
plt.xlabel('x [km]')
plt.ylabel('y [km]')
```
