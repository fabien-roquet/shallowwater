# Inertial motion and inertia–gravity waves

Experiment 15 illustrates how horizontal motion turns under Coriolis acceleration and how inertial motion connects to propagating shallow-water waves. These are integrations of the existing linear SW equations, with no forcing, drag or viscosity. The channel is periodic in x and closed in y.

For a pure inertial oscillation,
$$u_t=fv,\qquad v_t=-fu,\qquad T_f=2\pi/|f|.$$
Northern-hemisphere velocities rotate clockwise; southern-hemisphere velocities rotate anticlockwise. At finite wavenumber, shallow water supports the inertia–gravity (Poincaré) branch:
$$\omega^2=f^2+gHK^2,\qquad \omega/|f|=\sqrt{1+(KL_d)^2},\qquad L_d=\sqrt{gH}/|f|.$$
The low-wavenumber limit approaches inertial oscillation; at high wavenumber gravity dominates. This depth-averaged model illustrates horizontal inertial motion and inertia–gravity waves, rather than the vertical structure of three-dimensional inertial waves.

Here $H=100$ m and $|f|=10^{-4}$ s$^{-1}$ give $L_d=313$ km and $T_f=17.45$ h. All velocities are independent of x, so these runs isolate meridional wave dynamics.

## Analytic check

The standing cases begin with $v=V\sin(\pi y/L_y)$, $u=\eta=0$. This obeys impermeability and contains no initially balanced PV anomaly. On the continuous equations,
$$v=V\sin(\ell y)\cos(\omega t),\quad
u=\frac{fV}{\omega}\sin(\ell y)\sin(\omega t),\quad
\eta=-\frac{H\ell V}{\omega}\cos(\ell y)\sin(\omega t),\qquad\ell=\pi/L_y.$$

C-grid interpolation replaces $f$ in this modal oscillator by $f\cos(\ell\Delta y/2)$ and the pressure/divergence wavenumber by $2\sin(\ell\Delta y/2)/\Delta y$. The script and regression tests check the numerical solution against this discrete frequency. Metadata records energy change and height error. No analytical state replaces the time integration.
