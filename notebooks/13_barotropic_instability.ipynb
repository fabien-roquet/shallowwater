# Wind-driven jets and barotropic instability

A steady eastward **sech² wind stress** builds a zonal jet from rest in a channel that is periodic east–west. After 25 days we add a tiny balanced periodic disturbance and continue for 100 days under the same wind.

Compare strong wind with momentum advection enabled and disabled. The animation shows rotational velocity streamfunction above relative vorticity, with one column per case. Set `args.include_weak = True` before running to also compute the weak-wind diagnostics. All cases retain nonlinear mass transport and local-depth wind acceleration. The circulation is never held fixed.

**Predict:** Which jet will develop meanders? How should its eddy kinetic energy change? Can the PV-gradient criterion by itself establish instability?

## Inspect and interpret

For quantitative growth rates, change `quick` to `False` and compare resolutions. The advection-off case removes both terms that represent momentum advection in the vector-invariant equations; it is not the fully linear model. The seeded background continues to strengthen and evolve.

Inspect the diagnostic figure and the saved `.npz`/`.json` files in `animations/`. A growing disturbance, a mean PV-gradient reversal, and persistence under refinement support the instability interpretation.

Streamfunction uses `u_rot = -dψ/dy`, `v_rot = dψ/dx`. A periodic-channel Poisson solve recovers the rotational part of the velocity while retaining the net zonal flow. Its contours are streamlines of this component; the full shallow-water velocity can also diverge.
